以上的所有数据结构均包括两个特性：有序/无序，可变与不可变

有序：可以用前后双向下标索引访问序列的任意元素。 无序不可以了。
双向下标索引 ：正向下标 0 - n-1 , 由后向前的下标： -n -3 -2 -1  无序：不能使用下标访问元素。

可变与不可变：
可变：增、删、改，一个对象可变，在python里意味着，结构会附带一组函数处理增删。
不可变：查也有一组函数负责查找。

参数的类型：
（1）普通参数：位置对应的参数
（2）默认值参数：形参带有初始值，如果不传参，变量使用默认值，默认值参数的右侧不能有普通参数,默认值参数只初始化一次，初始化后转为函数的属性 可变对象做默认值的时候需要注意初始化问题。
（3）关键字参数：实参使用，带有形参名称且赋值的参数，打破位置对应。
（4）可变长度参数:*参数名  将多个/余的参数元组的形式放置在参数名下。
   **参数名  将多余关键字参数以字典的形式存储在参数名下
